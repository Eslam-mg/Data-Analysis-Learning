```python
import pandas as pd

df = pd.read_csv("DataCoSupplyChainDataset.csv", encoding="latin-1")

print(df.columns)
```

# Change column names

```python
df.rename(columns={
    "Type": "Payment",
    "Days for shipping (real)": "Real ship days",
    "Days for shipment (scheduled)": "Scheduled ship days",
    "Late_delivery_risk": "Is late delivery",
    "Order Item Total": "Sales after discount",
    "Order Profit Per Order": "Order Profit",
    "Order Item Product Price": "Order Item Price",
    "Product Card Id": "Product Id",
    "shipping date (DateOrders)": "Shipping Date",
    "order date (DateOrders)": "Order Date"
}, inplace=True)
df.columns
```

<!-- marge Customer Fname and Customer Lname to Customer Name -->

```python
df["Customer Name"] = df["Customer Fname"] + " " + df["Customer Lname"]
```

<!-- drop columns -->

```python
df = df.drop(columns=[
    'Benefit per order',
    'Sales per customer',
    'Customer Email',
    'Customer Fname',
    'Customer Lname',
    'Customer Password',
    'Order Item Cardprod Id',
    'Order Zipcode',
    'Product Description',
    'Product Status',
    'Order Customer Id',
    'Product Category Id',
    'Product Price',
])
df.shape
```

# change type of columns 

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Shipping Date'] = pd.to_datetime(df['Shipping Date'])
text_cols = [
    'Category Id', 'Customer Id', 'Order Id',
    'Product Id', 'Department Id',
    'Order Item Id',
]

for col in text_cols:
    df[col] = df[col].astype(str)
```

# extract dimension order table

```python
Order_Dim = [
    'Order Id',
    'Real ship days',
    'Scheduled ship days',
    'Is late delivery',
    'Market',
    'Order Country',
    'Order Region',
    'Order State',
    'Order City',
    'Order Date',
    'Order Status',
    'Payment',
    'Shipping Date',
    'Shipping Mode',
    'Delivery Status',
]
order_dim_df = df[Order_Dim]
order_dim_df.to_csv('Order_Dim.csv', index=False)
```

# extract dimension customer table

```python
Customer_Dim = [
    'Customer Id',
    'Customer Name',
    'Customer Segment',
    'Customer Country',
    'Customer State',
    'Customer City',
    'Customer Street',
    'Customer Zipcode',
    'Latitude',
    'Longitude',
]
customer_dim_df = df[Customer_Dim]
customer_dim_df.to_csv('Customer_Dim.csv', index=False)
```

# extract dimension product table

```python
Product_Dim = [
    'Product Id',
    'Product Name',
    'Category Id',
    'Category Name',
    'Department Id',
    'Department Name',
    'Product Image',
]
product_dim_df = df[Product_Dim]
product_dim_df.to_csv('Product_Dim.csv', index=False)
```

# extract fact table

```python
Fact_Table = [
    'Order Item Id',
    'Order Id',
    'Customer Id',
    'Product Id',
    'Order Item Quantity',
    'Order Item Discount',
    'Order Item Discount Rate',
    'Sales',
    'Sales after discount',
    'Order Profit',
    'Order Item Profit Ratio',
    'Order Item Price',
]
fact_table_df = df[Fact_Table]
fact_table_df.to_csv('Fact_Table.csv', index=False)
```

# extract dimension date table

```python
min_date = df['Order Date'].min()
max_date = df['Order Date'].max()
Date_Dim = pd.DataFrame({
    'Date': pd.date_range(start=min_date, end=max_date, freq='D')
})

Date_Dim['Year'] = Date_Dim['Date'].dt.year
Date_Dim['Month'] = Date_Dim['Date'].dt.month
Date_Dim['Month Name'] = Date_Dim['Date'].dt.strftime('%B')
Date_Dim['Month Short'] = Date_Dim['Date'].dt.strftime('%b')
Date_Dim['Quarter'] = 'Q' + Date_Dim['Date'].dt.quarter.astype(str)
Date_Dim['Day'] = Date_Dim['Date'].dt.day
Date_Dim['Day Name'] = Date_Dim['Date'].dt.strftime('%A')
Date_Dim['Day Short'] = Date_Dim['Date'].dt.strftime('%a')
Date_Dim['Week Number'] = Date_Dim['Date'].dt.isocalendar().week


Date_Dim.to_csv('Date_Dim.csv', index=False)
```
